```python
#Customer detail
age = 30
salary = 55000.0
loan_amount = 200000
existing_loans = 2
credit_score = 720
is_citizen = True

# --- Arithmetic Operators ---
emi = loan_amount / 24 #Monthly installment
total_income_yearly = salary * 12
remaining_Salary = total_income_yearly - (emi * 12)

print("Monthly EMI:", emi)
print("Yearly Income:", total_income_yearly)
print("Remaining Salary after EMI:", remaining_Salary)

# --- Comparison Operators ---
print("Is credit score good?", credit_score >= 700)
print("Is EMI less than 50% of salary?", emi < (salary * 0.5))
print("Is age less than 60%?", age < 60)

# --- Logical Operators ---
eligible = (credit_score >= 700) and (emi < (salary * 0.5)) and (age < 60)
print("Loan Eligibility:", eligible)
```

```python
# Function  to issue books
def issue_books(student_name, books):
    print(f"\n {student_name} borrowed:")
    for book in books:
        print("-", book)

# Funnction to calculated fine
def calculated_fine(days_late):
    fine_per_day = 5
    return days_late * fine_per_day

# Function to generate report
def student_report(student_name, books, days_late):
    issue_books(student_name, books)
    fine = calculated_fine(days_late)
    print(f"Fine to be paid: ${fine}\n")

# Example usage
student_report("Rahul", ["Python Basics", "Data Science 101"], 3)
student_report("Priya", ["Machine Learning"],0)
```
