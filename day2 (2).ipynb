```python
import pandas as pd
import numpy as np

# Create a sample DataFrame with the expected columns and some missing values
df = pd.DataFrame({
    'Age': [25, 30, 35, 40, 45],
    'Salary': [50000, 60000, 70000, 80000, 90000],
    'Math_Score': [85, 90, np.nan, 75, 88],
    'Writing_Score': [70, np.nan, 80, 92, 78],
    'Parental_Education': ['some college', 'high school', 'bachelor\'s degree', np.nan, 'some college'],
    'Reading_Score': [72, 85, 88, np.nan, 90],
    'Attendance': ['P', 'A', 'P', 'P', np.nan]
})

# 1. Fill with Mean (Numerical)
df['Math_Score'] = df['Math_Score'].fillna(df['Math_Score'].mean())
# 2. Fill with Median
df['Writing_Score'] = df['Writing_Score'].fillna(df['Writing_Score'].median())
# 3. Fill with Mode (Categorical)
df['Parental_Education'] = df['Parental_Education'].fillna(df['Parental_Education'].mode()[0])
# 4. Forward Fill / Backward Fill
df['Reading_Score'] = df['Reading_Score'].ffill()
# 5. Drop rows (if too many missing)
df = df.dropna(subset=['Attendance'])

print("DataFrame after cleaning:\n", df)
```

```python
# Total Score
df['Total_Score'] = df[['Math_Score','Reading_Score','Writing_Score']].sum(axis=1)
# Average Score
df['Average_Score'] = df['Total_Score'] / 3
# Performance Category
df['Performance_Level'] = pd.cut(
    df['Average_Score'],
    bins=[0,60,75,100],
    labels=['Low','Medium','High']
)
# Attendance Category
# Convert 'P' and 'A' to numerical values for binning
df['Attendance_Numeric'] = df['Attendance'].map({'P': 95, 'A': 50}) # Example mapping
df['Attendance_Level'] = pd.cut(
    df['Attendance_Numeric'],
    bins=[0,75,90,100],
    labels=['Poor','Average','Good']
)
```

```python
# Display available columns for debugging
print("DataFrame columns:", df.columns)

# 1. Average performance by gender
# print(df.groupby('Gender')['Average_Score'].mean()) # This line caused KeyError as 'Gender' column does not exist
# 2. Distribution of performance levels
print(df['Performance_Level'].value_counts())
# 3. Correlation between attendance & average score
print(df[['Attendance_Numeric','Average_Score']].corr())
# 4. Parental education impact
print(df.groupby('Parental_Education')['Average_Score'].mean())
```
